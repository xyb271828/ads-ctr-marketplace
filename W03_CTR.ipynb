# Click prediction and calibration
Install `requirements-criteo.txt` and point the benchmark at the folder containing the five dated Parquet files.
The earlier three dates are used for training, the fourth for validation, and the fifth for test.


```python
from pathlib import Path
from ads.ctr import benchmark

# Use data/criteo for the repository download or the local folder shown below.
source = Path('data/criteo')
if not source.is_dir():
    source = Path('~/data/criteo').expanduser()
result = benchmark(source, 'runs/criteo-notebook', rows=200_000, full=True)
[(name, m['log_loss'], m['auc'], m['ne'], m['ece']) for name, m in result['models'].items()]
```

```python
import matplotlib.pyplot as plt

for name in ("hashed_lr", "platt", "isotonic", "negative_sampled", "negative_corrected"):
    points = result["models"][name]["reliability"]
    plt.plot([r["predicted"] for r in points], [r["observed"] for r in points], "o-", label=name)
plt.plot([0, 1], [0, 1], "--", color="gray")
plt.xlabel("Predicted CTR")
plt.ylabel("Observed CTR")
plt.legend()
```
