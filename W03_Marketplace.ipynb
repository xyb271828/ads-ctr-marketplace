# GSP auction and budget pacing
The paired replay holds traffic and click draws fixed across policies.


```python
from ads.marketplace import experiment

result = experiment()
result["summary"]
```

```python
import matplotlib.pyplot as plt
import numpy as np

for variant, policies in result["replays"].items():
    for paced, days in policies.items():
        spend = np.mean([[hour["spend"] for hour in day["hourly"]] for day in days], axis=0)
        plt.plot(range(24), np.cumsum(spend), label=f"{variant}, paced={paced}")
plt.ylabel("Cumulative spend")
plt.xlabel("Simulated hour")
plt.legend()
```
